Simple TLU Perceptron on iris dataset

```python
from sklearn.datasets import load_iris
import numpy as np
from sklearn.linear_model import Perceptron
```

```python
dataset = load_iris()
```

```python
dataset.keys()
```

```python
x = dataset.data
y= dataset.target
```

```python
x = x[:,(2,3)]
y = (y == 0).astype(int)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.9)
```

```python
prd_clf= Perceptron()
```

```python
from sklearn.model_selection import cross_val_score
score = cross_val_score(prd_clf,x_train,y_train,cv=5,n_jobs=-1)
```
